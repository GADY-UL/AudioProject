```python
import IPython.display as ipd
checking = 'D:\\codethings\\GADy\\AudioMINST\\processed_data\\smol_test'
before = checking + '\\' + 'before_0_02_12.wav'
after = checking + '\\' + 'after_0_02_12.wav'
ipd.Audio(before)
```

```python
ipd.Audio(after)
```

```python
import librosa
import matplotlib.pyplot as plt
import soundfile as sf
l_before, sr_before = librosa.load(before, sr=None)

amp_before = librosa.stft(l_before)
db_before = librosa.amplitude_to_db(abs(amp_before))

librosa.display.specshow(
    db_before,
    sr=sr_before,
    x_axis='time',
    y_axis='hz'
)
plt.colorbar()
print(sf.info(before))
```

```python
l_after, sr_after = librosa.load(after, sr=None)

amp_after = librosa.stft(l_after)
db_after = librosa.amplitude_to_db(abs(amp_after))

librosa.display.specshow(
    db_after,
    sr=sr_after,
    x_axis='time',
    y_axis='hz'
)
plt.colorbar()
print(sf.info(after))
```

```python
import os
import librosa
import soundfile as sf
import noisereduce as nr

data_path = 'D:\\codethings\\GADy\\AudioMINST\\data'
clean_path = 'D:\\codethings\\GADy\\AudioMINST\\processed_data'

def nreduce_downsample(path):
    orig_filename = path.split('\\')[-1]
    orig_foldername = path.split('\\')[-2]
    target_sr = 16000
    create_path = clean_path + '\\' + orig_foldername + '\\' + orig_filename
    loaded_file, sr = librosa.load(path, sr=None)
    resampled_file = librosa.resample(y=loaded_file, target_sr=target_sr, orig_sr=sr)
    reduced_resampled = nr.reduce_noise(resampled_file, sr=target_sr)
    if not os.path.exists(clean_path + '\\' + orig_foldername):
        os.makedirs(clean_path + '\\' + orig_foldername)
    sf.write(create_path, reduced_resampled, target_sr)
    # print(create_path)
    
for folder in os.listdir(data_path): # iterate through folders
    dir_path = os.path.join(data_path, folder) # create path as var
    if os.path.isdir(dir_path): # if the path is a folder
        for file in os.listdir(dir_path): # iterate through files
            file_path = os.path.join(dir_path, file) # create path as var
            if os.path.isfile(file_path): # if file on path is a file
                nreduce_downsample(file_path)
```

```python
test_path = 'D:\\codethings\\GADy\\AudioMINST\\data\\60\\9_60_6.wav'
nreduce_downsample(test_path)
```
